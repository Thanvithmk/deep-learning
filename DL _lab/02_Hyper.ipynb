```python
from keras.layers import Dense,Flatten,Input
from keras.datasets import cifar10
from keras.models import Sequential
from keras.utils import to_categorical

(x_train,y_train),(x_test,y_test)=cifar10.load_data()
print(y_train.shape)

x_train,x_test=x_train/255,x_test/255
y_train=to_categorical(y_train)
y_test=to_categorical(y_test)

def train_model(act,opt):
    model=Sequential()
    model.add(Input(shape=(32, 32, 3)))
    model.add(Flatten())
    model.add(Dense(512,activation=act))
    model.add(Dense(10,activation='softmax'))
    model.compile(optimizer=opt,loss='categorical_crossentropy',metrics=['accuracy'])
    return model

activations=["tanh","relu","sigmoid"]
optimizers=["adam","sgd"]
result = {}
for a in activations:
    for o in optimizers:
        model=train_model(a,o)
        hist=model.fit(x_train,y_train,epochs=100,verbose=0)
        loss,accuracy=model.evaluate(x_test,y_test,verbose=0)
        result[(a,o)]=accuracy

best_model=max(result,key=result.get)
print(best_model)


print("Accuracy of all models:")
for k, v in result.items():
    print(f"{k}: {v:.4f}")

print("\nBest Model:", best_model)
print("Best Accuracy:", result[best_model])
```
