```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import InputLayer, Conv1D, MaxPooling1D, Flatten, Dense

data = pd.read_csv("sample.csv")

X = data.iloc[:, :-1].values

scaler = MinMaxScaler()
X = scaler.fit_transform(X)

def create_sequences(data, window):
    X_seq = []
    y_seq = []
    for i in range(len(data) - window):
        X_seq.append(data[i:i + window])
        y_seq.append(data[i + window])  # No index! Grabs [f1, f2, f3]
    return np.array(X_seq), np.array(y_seq)

window = 7
X, y = create_sequences(X, window)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

model = Sequential()
model.add(InputLayer(shape=(X_train.shape[1], X_train.shape[2])))
model.add(Conv1D(filters=64,kernel_size=2,activation='relu'))
model.add(MaxPooling1D(pool_size=2))
model.add(Flatten())
model.add(Dense(50, activation='relu'))
model.add(Dense(3))

model.compile(optimizer='adam', loss='mse')
model.fit(X_train,y_train,epochs=50,batch_size=8)

pred = model.predict(X_test)
actual_values = scaler.inverse_transform(pred)
print(actual_values)
```

| Problem Type | Target Data Format ($y$) | Recommended Loss | When to Use | Final Layer Setup |
| :--- | :--- | :--- | :--- | :--- |
| **Regression** | Continuous numbers (e.g., `12.5`, `340.2`) | `'mse'` (Mean Squared Error) | Default choice for continuous values; penalizes large errors heavily. | `Dense(n)`, no activation |
| **Regression** | Continuous numbers with heavy noise/outliers | `'mae'` (Mean Absolute Error) | Use when your dataset contains extreme anomalies or outliers. | `Dense(n)`, no activation |
| **Binary Classification** | Two options (e.g., `0` or `1`, True or False) | `'binary_crossentropy'` | Predicting one of two classes (e.g., Spam vs. Not Spam). | `Dense(1)`, activation=`'sigmoid'` |
| **Multi-Class Classification** | One-hot encoded arrays (e.g., `[1, 0, 0]`, `[0, 1, 0]`) | `'categorical_crossentropy'` | Predicting 3+ categories when labels are formatted as vectors. | `Dense(num_classes)`, activation=`'softmax'` |
| **Multi-Class Classification** | Integers (e.g., `0`, `1`, `2`, `3`) | `'sparse_categorical_crossentropy'` | Predicting 3+ categories when labels are single integers (saves memory). | `Dense(num_classes)`, activation=`'softmax'` |
